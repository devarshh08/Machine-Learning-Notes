### Tensor operations and gradients

We combine tensors with the usual arithmetic operations.

```python
x = torch.tensor(3.)
w = torch.tensor(4., requires_grad=True)
b = torch.tensor(5., requires_grad=True)
```

```python
#combining these tensors to create a new one

y = w * x + b
y
```

```python
#computing derivatives
y.backward()
```

```python
#displaying gradients

print('dy/dx:', x.grad)
print('dy/dw:', w.grad)
print('dy/db:', b.grad)
```

Gradient is used when dealing with matrices
&
Derivative is used when dealing with numbers

### Interoperability with numpy

Numpy is used for scientific computing. It is efficient on operating on large multi-dimensional arrays(efficient because it is implemented in c++), and has a large ecosystem of supporting libraries:

- Matplotlib : plotting and visualization
- OpenCV : Image and video processing
- Pandas : file I/O and data analysis

```python
import numpy as np
```

```python
x = np.array([[1, 2], [3, 4.]])
x
```

```python
#we can convert numpy array to pytorch tensor

y = torch.from_numpy(x)
y
```

```python
#we can also convert pytorch tensor to numpy

z = y.numpy()
z
```

Numpy and pytorch exhibit the same properties, but we use Pytorch because it is optimized to be used on gpus. 

Pytorch is optimized to perform on gpus and numpy arrays run on cpus. So pytorch is faster than numpy.
